# Data Collection

This is where the raw data for analysis is collected. I do not own the raw data.

The data is collected from `nflreadpy` which can be found here https://nflreadpy.nflverse.com/


This saves the schedules and team stats for all the teams of the 2007 season. 

```python
from pathlib import Path
import nflreadpy as nfl
SEASON = 2007
RAW_DIR = Path("..") / "data"/ "raw"
```

```python
schedules = nfl.load_schedules(SEASON)
schedules.head()
```

```python
schedules.shape
```

```python
team_stats = nfl.load_team_stats(SEASON)
```

```python
team_stats.head()
```

```python
team_stats.shape
```

```python
schedules.write_parquet(RAW_DIR / f"schedules__{SEASON}.parquet")
team_stats.write_parquet(RAW_DIR / f"team_stats__{SEASON}.parquet")
```
